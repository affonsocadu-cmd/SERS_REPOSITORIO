8. **Explique por que esse segundo conjunto de registros pode merecer maior atenção da equipe de energia.**

O DataFrame `df_alto_consumo_fp_baixo` merece maior atenção da equipe de energia porque ele identifica os períodos em que a indústria está enfrentando **alto consumo de energia (Consumo_kWh)** e, simultaneamente, um **fator de potência atrasada abaixo do limite estabelecido**. Essa combinação é crítica por várias razões:

1.  **Custos Elevados**: Alto consumo de energia naturalmente implica em custos operacionais mais altos. Quando isso ocorre junto com um fator de potência baixo, os custos podem ser ainda maiores devido a:
    *   **Perdas na Rede**: Um fator de potência baixo significa que uma corrente elétrica maior é necessária para fornecer a mesma quantidade de potência ativa (útil). Isso causa mais perdas por aquecimento nos condutores e transformadores.
    *   **Multas ou Penalidades**: Muitas concessionárias de energia aplicam multas a consumidores industriais que operam com fator de potência abaixo de um determinado limite (geralmente 0,92 ou 0,95), pois isso sobrecarrega a rede de distribuição.

2.  **Ineficiência Operacional**: Um fator de potência baixo indica que a energia reativa é excessiva em relação à energia ativa, o que representa um uso ineficiente da eletricidade. A equipe de energia pode investigar a causa desse baixo fator de potência (por exemplo, excesso de motores indutivos sem compensação) e implementar soluções.

3.  **Potencial de Otimização**: A identificação desses períodos específicos permite que a equipe concentre seus esforços de otimização. Ao invés de tentar melhorar a eficiência em toda a operação, eles podem focar nos horários ou condições que mais contribuem para o desperdício de energia e custos adicionais. Isso pode incluir a instalação de bancos de capacitores para correção do fator de potência ou ajustes no agendamento de cargas.

Em resumo, esses registros apontam para as situações mais problemáticas em termos de eficiência energética e custos, tornando-os prioritários para análise e intervenção da equipe de gestão de energia.

7. **Explique por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes sem antes observar a escala de cada variável.**

### **Explicação Técnica**

**Por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes sem antes observar a escala de cada variável?**

Não é adequado aplicar um limite absoluto fixo (ex: um valor arbitrário em MW ou MWh) para ambas as fontes porque a energia solar e a energia eólica possuem **perfis de geração, fatores de capacidade e capacidades instaladas totalmente distintos**:

1. **Intermitência e Ciclo Diurno:** A energia solar opera dentro de uma janela restrita (horas de sol), atingindo picos concentrados perto do meio-dia e caindo a zero durante a noite. A energia eólica, por sua vez, pode gerar continuamente ao longo das 24 horas, variando de acordo com regimes de ventos regionais.
2. **Diferenças de Escala e Capacidade:** Se uma usina ou parque eólico tiver uma capacidade máxima pico instalada significativamente maior (ou menor) do que o parque solar, utilizar um único corte numérico distorceria a análise. Um valor alto demais poderia fazer parecer que a fonte de menor escala nunca atinge picos de alta eficiência, enquanto um valor baixo demais classificaria quase toda a produção da fonte de maior escala como 'alta'.

Ao normalizar pelo limite relativo de 70% da capacidade máxima individual de cada fonte, avalia-se o comportamento operacional e o desempenho relativo de cada tecnologia em relação ao seu próprio potencial.
