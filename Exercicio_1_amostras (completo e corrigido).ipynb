# 2. Renomeie Appliances para Consumo_Eletrodomesticos e simplifique o nome de pelo menos três atributos ambientais.

```python
dados.columns
```

```python
dados = dados.rename({'Appliances': 'Consumo_eletrodomesticos',
             'lights':'Luminosidade',
              'T1':'Temperatura_1',
              'RH_1': 'Umidade_1',
              'T2':'Temperatura_2',
              'RH_2':'Umidade_2',
              'T3':'Temperatura_3',
              'RH_3':'Umidade_3' }, axis=1)

print(dados.columns)
```

# 3. Determine o maior consumo de eletrodomésticos registrado na amostra.

```python
CMAX = dados ["Consumo_eletrodomesticos"].max()
print(f'O consumm máximo de eletrodomésticos foi de {CMAX:.2f} KW')
```

# 4. Calcule um limiar correspondente a 70% do valor máximo e crie um DataFrame contendo somente os registros acima desse limite.  

```python
df_alto_consumo = dados[dados['Consumo_eletrodomesticos'] > C70]
print(f'O limiar correspondente a 70% é de {C70:.2f} KW')
```

#5. Conte os registros selecionados e calcule qual percentual da amostra eles representam.

```python
qtd_registros = len(df_alto_consumo)
percentual = (qtd_registros / len(dados)) * 100

print(f"Registros com consumo acima de 70%: {qtd_registros}")
print(f"Percentual da amostra: {percentual:.2f}%")
```

# 6. Calcule a temperatura média de T1 e crie outro DataFrame contendo simultaneamente consumo acima de 70% do máximo e temperatura acima da média.

```python
temp_media_t1 = dados['Temperatura_1'].mean()
df_consumo_e_temp = dados[
    (dados['Consumo_eletrodomesticos'] > C70) &
    (dados['Temperatura_1'] > temp_media_t1)
]

print(f"Média da Temperatura_1 (T1): {temp_media_t1:.2f}°C")
print(f"Registros com consumo > 70% E temp > média: {len(df_consumo_e_temp)}")
```

# 7. Compare os dois DataFrames e explique o efeito da inclusão da temperatura como segundo critério.

```python
print("\n--- Comparação entre os DataFrames ---")
print(f"DataFrame Ex. 4 (Apenas Consumo Elevado): {len(df_alto_consumo)} registros")
print(f"DataFrame Ex. 6 (Consumo Elevado + Temp1 > Média): {len(df_consumo_e_temp)} registros")
```
